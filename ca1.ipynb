```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
uploaded = files.upload()

filename = next(iter(uploaded))
print("Uploaded file:", filename)
```

```python
# Read the image
img = cv2.imread(filename)

# Check if image was read successfully
if img is None:
    print("Error: Image could not be loaded.")
else:
    print("Image loaded successfully!")

# Print image shape
print("Original Image Shape:", img.shape)

# Convert BGR to RGB for displaying with Matplotlib
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Display original image
plt.figure(figsize=(8, 6))
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
# Convert BGR image to RGBA
img_rgba = cv2.cvtColor(img, cv2.COLOR_BGR2RGBA)

# Convert BGR image to Grayscale
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Print shapes
print("Original Image Shape (BGR):", img.shape)
print("RGBA Image Shape:", img_rgba.shape)
print("Grayscale Image Shape:", img_gray.shape)

# Display RGBA and Grayscale images
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgba)
plt.title("RGBA Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(img_gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.show()
```

```python
# Convert BGR image to RGB
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Separate RGB channels
red = rgb[:, :, 0]
green = rgb[:, :, 1]
blue = rgb[:, :, 2]

# Display individual channels
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(red, cmap="gray")
plt.title("Red Channel")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(green, cmap="gray")
plt.title("Green Channel")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(blue, cmap="gray")
plt.title("Blue Channel")
plt.axis("off")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(red.ravel(), bins=256, range=[0, 256], alpha=0.5, label="Red")
plt.hist(green.ravel(), bins=256, range=[0, 256], alpha=0.5, label="Green")
plt.hist(blue.ravel(), bins=256, range=[0, 256], alpha=0.5, label="Blue")

plt.title("Histogram of RGB Channels")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
plt.legend()

plt.savefig("rgb_histogram.png")
plt.show()
```

```python
# Apply histogram equalization to each RGB channel

red_eq = cv2.equalizeHist(red)
green_eq = cv2.equalizeHist(green)
blue_eq = cv2.equalizeHist(blue)

# Merge the equalized channels
enhanced_img = cv2.merge([red_eq, green_eq, blue_eq])

# Save the enhanced image
cv2.imwrite(
    "enhanced_image.png",
    cv2.cvtColor(enhanced_img, cv2.COLOR_RGB2BGR)
)

print("Histogram equalization applied successfully.")
```

```python
# Display original and enhanced images

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(enhanced_img)
plt.title("Enhanced Image")
plt.axis("off")

plt.savefig("original_vs_enhanced.png")
plt.show()
```

```python
# Save individual RGB channel images

cv2.imwrite("red_channel.png", red)
cv2.imwrite("green_channel.png", green)
cv2.imwrite("blue_channel.png", blue)

print("All generated images have been saved.")
```
